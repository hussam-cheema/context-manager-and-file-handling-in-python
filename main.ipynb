```python
file = open("fruits.txt", "w")

fruits = ['apple', 'orange', 'kiwi']

for f in fruits:
    file.write(f + '\n')

file.close()
```

```python
file = open("my_file.txt", "w")

file.write("Hello Hussam\nHey, Welcome to my youtube channel")

file.close()
```

```python
file = open("my_file.txt", "r")

print(file.read())

file.seek(0)

print(file.readline())

file.close()
```

```python
file = open("my_file.txt", "r")

for line in file:
    print(line)

file.close()
```

```python
file = open("my_file.txt", "a")

file.write("\nThis is my 3rd message")

file.close()
```

```python
with open("data.txt", "w") as file:
    file.write("Hello World")
```

```python
import json

data = {"name": "Hussam", "age": 29}

with open("data.json", "w") as file:
    json.dump(data, file)
```

```python
with open("data.json", "r") as file:
    result = json.load(file)
```
